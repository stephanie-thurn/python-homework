# Conditions and Loops

* Given: Two positive integers a and b (a<b<10000).

* Return: The sum of all odd integers from a through b, inclusively.

* Hint: You can use a % 2 == 1 to test if a is odd.

```python
a = 4787
b = 8938
```

```python
total_sum = 0
```

```python
for num in range(a, b +1):
    if num % 2 ==1:
        total_sum +=num
```

```python
print(total_sum)
```
