# Counting DNA Nucleotides
## Exercise
* A string is simply an ordered collection of symbols selected from some alphabet and formed into a word; the length of a string is the number of symbols that it contains.

* An example of a length 21 DNA string (whose alphabet contains the symbols 'A', 'C', 'G', and 'T') is "ATGCTTCAGAAAGGTCTTACG."

* Given: A DNA string s of length at most 1000 nt.

* Return: Four integers (separated by spaces) counting the respective number of times that the symbols 'A', 'C', 'G', and 'T' occur in s.



```python
sample_string = "CGCTGCCACTTTCATGGGCACTGGACTTATATTGGCAGACCAGCCACAGAAAACTAACATGTAGACAAAGCCGGGAGCTCCGCGATCGTTGTTTTTAGTCAACGGACAACGCCACGAACCCCGCCACACTAGATGATATGCTCGCTAGAAAGTTTACCCAACAGGCAGAAGCCCATAACAGGTTTTACGTTTTGTAGGCCATGAGCGGGCTGTTTAGATAAGCACCCCTGCTGCCAGGTATACGCCAGCTGAAAGGCATCTCCTAATGCTGAGACGTTTACCATGACTTCCGAGCCGGATCTCATTGTAGATTGTTCTGAACTGATGGCACTAACATTGGACACGCGTGCTAAGGCTTTCTTATTGAGCTTAATAGTTCGATTCTATTTCGAAAGTTTCCGGAAAATTACCCGTATTTATCTACCATCCGTAATGCGACGAATTAACATCTGGGTCTCAGCCGGGCCTTCAAAAGGTCCAGTCAGCGACGCAAGACGTCATCGCCCGTACAGGATAACGACAGTTAAGCGTGTCTTCGTAAGGTTGGAATGCAATTGACCAACATTAAGGGCGTCCGGCAACGAGACTAATTGCCAGGTTCATCTCTTTCGTTCGGTCATGTCCCTGGAACTCATCAGGGCTGGTAAATAGGACACGCCGTGTATGAGAGCTGGACATGTGGACATTAATTCTATGCAAGTCAGCCGTGGCCATTGTGTCTGGCAAGGTAGGTGCGCTTTTTCCCATGGCGCAGAGGCTACCAATCTTTCGCGGGACCCGCCACTCCTACGCGAGCGATGCAACGTTTGCACCAGCACGCGAAGGGGAGGTATGTATCGAAGATCGCTAAAGAAGACCCCTGTAGAAACACCAAGACTCCACACATAAGGCGGGCCAATGTACTCTGGGAATTTCTAAATACCCGAGCCTCACTTCCCCCTGCCTACCAGTATCGAAACCAAAT"
```

```python
count_A = sample_string.count("A")
```

```python
count_C = sample_string.count("C")
```

```python
count_G = sample_string.count("G")
```

```python
count_T = sample_string.count("T")
```

```python
print(count_A, count_C, count_G, count_T)
```
